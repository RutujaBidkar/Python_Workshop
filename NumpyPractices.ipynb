```python
#Create a 2D array
a = np.array([[1,2,3],
              [4,5,6]])
a
```

```python
#Check the Shape
print(a.shape)
```

```python
#Check the Number of Dimensions
print(a.ndim)
```

```python
#Check the Total Number of Elements
print(a.size)
```

```python
#find the maximum value
print(a.max())
```

```python
#find minimum value
print(a.min())
```

```python
#Find the Sum of All Elements
print(a.sum())
```

```python
#Mean calculate
print(np.mean(a))
```

```python
#Find the stdard deviation
print(np.std(a))
```

```python
#Create an Array of Zeros
print(np.zeros(5))
```

```python
#Create an array of ones
print(np.ones(5))
```

```python
#Create an Array with a Range of Numbers
print(np.arange(0,10))
```
